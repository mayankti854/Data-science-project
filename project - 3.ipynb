```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from statsmodels.tsa.arima.model import ARIMA
from sklearn.metrics import mean_absolute_error, mean_squared_error
```

```python
df = pd.read_csv("stock_data.csv")
df.head()
```

```python
df = pd.read_csv("stock_data.csv")

df["Date"] = pd.to_datetime(df["Date"])
df = df.sort_values("Date")

print(df.head())
print(df.info())
```

```python
. Plot stock trend + moving averages
```

```python
df["MA_5"] = df["Close"].rolling(window=5).mean()
df["MA_10"] = df["Close"].rolling(window=10).mean()

plt.figure(figsize=(12, 5))

plt.plot(df["Date"], df["Close"], label="Close Price")
plt.plot(df["Date"], df["MA_5"], label="5-Day Moving Average")
plt.plot(df["Date"], df["MA_10"], label="10-Day Moving Average")

plt.xlabel("Date")
plt.ylabel("Price")
plt.title("Stock Price Trend")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# 4. TRAIN-TEST SPLIT
```

```python
prices = df["Close"].astype(float)

train = prices.iloc[:-10]
test = prices.iloc[-10:]

print("Training data:")
print(train)

print("\nTesting data:")
print(test)
```

```python
train = train.astype(float)
train = train.replace([np.inf, -np.inf], np.nan)
train = train.dropna()
```

```python
# 5. ARIMA MODEL
```

```python
model = ARIMA(train, order=(1, 1, 1))

model_fit = model.fit()

print(model_fit.summary())
```

```python
# 6. FORECAST NEXT 30 DAYS
```

```python
forecast = model_fit.forecast(steps=len(test))

print("Actual values:")
print(test.values)

print("\nForecast values:")
print(forecast.values)
```

```python
# 7. CREATE COMPARISON TABLE
```

```python
comparison = pd.DataFrame({
    "Actual": test.values,
    "Forecast": forecast.values
})

print(comparison)
```

```python
8. Calculate Accuracy Metrics
```

```python
mae = mean_absolute_error(test, forecast)

rmse = np.sqrt(mean_squared_error(test, forecast))

print("MAE:", mae)
print("RMSE:", rmse)
```

```python
9. Plot Actual vs Forecast
```

```python
plt.figure(figsize=(12, 5))

plt.plot(test.index, test.values, marker="o", label="Actual")
plt.plot(test.index, forecast.values, marker="o", label="Forecast")

plt.xlabel("Day")
plt.ylabel("Close Price")
plt.title("ARIMA: Actual vs Forecast")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# 10. Finally — Forecast next 30 days
```

```python
final_model = ARIMA(prices, order=(1, 1, 1))

final_model_fit = final_model.fit()

future_forecast = final_model_fit.forecast(steps=30)

print("Next 30 Days Forecast:")
print(future_forecast)
```

```python
# 11. Create future dates
```

```python
future_dates = pd.date_range(
    start=df["Date"].iloc[-1] + pd.Timedelta(days=1),
    periods=30,
    freq="D"
)

future_df = pd.DataFrame({
    "Date": future_dates,
    "Forecast_Close": future_forecast.values
})

print(future_df)
```

```python
12. Plot the 30 - da
```

```python
plt.figure(figsize=(12, 5))

plt.plot(df["Date"], df["Close"], label="Historical Close")
plt.plot(
    future_df["Date"],
    future_df["Forecast_Close"],
    marker="o",
    label="30-Day Forecast"
)

plt.xlabel("Date")
plt.ylabel("Close Price")
plt.title("Stock Price — 30 Day ARIMA Forecast")
plt.legend()
plt.grid(True)
plt.show()
```
